```python
import numpy as np
import matplotlib.pyplot as plt

# Generate 100 random numbers (mean = 50, standard deviation = 10)
data = np.random.normal(loc=50, scale=10, size=100)

# Plotting the bell curve (histogram)
plt.hist(data, bins=15, color='skyblue', edgecolor='black')
plt.title("Normal Distribution (Mean=50, Std=10)")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

# question 2

```python
import numpy as np
from scipy import stats

# Array of 20 daily step counts
step_counts = [4800, 5200, 5000, 5100, 4900, 5000, 5300, 4700, 5000, 5100, 
               4800, 5050, 4950, 5150, 4850, 5000, 5250, 4750, 4900, 5100]

mean_val = np.mean(step_counts)
median_val = np.median(step_counts)
# stats.mode returns an object; we extract the actual mode value from it
mode_val = stats.mode(step_counts, keepdims=True)[0][0]

print(f"Mean: {mean_val:.1f}")
print(f"Median: {median_val}")
print(f"Mode: {mode_val}")

print("\nConclusion: The mean, median, and mode are all very close to 5000. Because they are approximately equal, this data perfectly reflects the shape of a normal distribution.")
```

# question 3

```python
import numpy as np

# Simulate 200 player scores (mean = 70, std = 15)
scores = np.random.normal(loc=70, scale=15, size=200)

# Count how many scores fall between 55 and 85
within_one_std = np.sum((scores >= 55) & (scores <= 85))
percentage = (within_one_std / 200) * 100

print(f"Total Players: 200")
print(f"Players scoring between 55 and 85: {within_one_std}")
print(f"Percentage: {percentage:.1f}%")

print("\nExplanation:")
print("According to the 68-95-99 rule, approximately 68% of values in a normal distribution fall within one standard deviation of the mean. Your result should be very close to 68% (around 136 players)!")
```
