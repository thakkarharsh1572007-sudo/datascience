# question 1

```python
def calculate_probability(event_count, total_outcomes):
    return event_count / total_outcomes

heads_prob = calculate_probability(1, 2)

print(f"The probability of getting 'Heads' is: {heads_prob} (or {heads_prob * 100}%)")
```

# question 2

```python
import random

total_rolls = 100
six_count = 0

for _ in range(total_rolls):
    if random.randint(1, 6) == 6:
        six_count += 1
        
experimental_probability = six_count / total_rolls

print(f"Total Rolls: {total_rolls}")
print(f"Number of times '6' appeared: {six_count}")
print(f"Experimental Probability of getting a '6': {experimental_probability:.2f}")
```

```python
recent_genres = [
    'pop', 'rock', 'pop', 'hiphop', 'jazz', 'pop', 'classical', 'rock', 'pop', 'indie',
    'pop', 'metal', 'hiphop', 'pop', 'country', 'pop', 'rock', 'blues', 'pop', 'edm'
]

genre_count = recent_genres.count('pop')
total_songs = len(recent_genres)

probability = genre_count / total_songs

print(f"Total recent songs: {total_songs}")
print(f"'pop' count: {genre_count}")
print(f"Probability of picking a 'pop' song: {probability:.2f}")
```

# question 4

```python
def prob_food_and_dessert(food_users, dessert_users):
    food_set = set(food_users)
    dessert_set = set(dessert_users)
    
    ordered_both = food_set.intersection(dessert_set)
    total_unique = food_set.union(dessert_set)
    
    return len(ordered_both) / len(total_unique)

food_order_users = ['u1', 'u2', 'u3', 'u4', 'u5']
dessert_order_users = ['u3', 'u4', 'u6', 'u7']

prob = prob_food_and_dessert(food_order_users, dessert_order_users)

print(f"Probability that a random user ordered both food and dessert: {prob:.2f}")
```

```python
print("Explanation: Conditional probability looks at the likelihood of Event B (buying headphones) given that Event A (buying a phone) already happened. If we divide the number of people who bought BOTH by the total number of people who bought the PHONE, we get this probability. If the probability is high, Flipkart's app can automatically trigger a 'Frequently Bought Together' popup.")

total_phone_buyers = 5000 
bought_phone_and_headphones = 1200 

prob_headphones_given_phone = bought_phone_and_headphones / total_phone_buyers

print(f"\nP(Headphones | Phone) = {prob_headphones_given_phone:.2f}")
if prob_headphones_given_phone > 0.20:
    print("Action: Trigger headphone recommendation popup on the checkout page.")
```
