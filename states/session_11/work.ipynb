# question 3

```python
import numpy as np

# 3 days (rows) x 4 dishes (columns)
orders_matrix = np.array([
    [10, 15, 8, 20],  # Day 1
    [12, 18, 5, 22],  # Day 2
    [9,  20, 10, 25]  # Day 3
])

dish_totals = np.sum(orders_matrix, axis=0)

print("--- Total Orders Per Dish Across 3 Days ---")
for i, total in enumerate(dish_totals):
    print(f"Dish {i+1}: {total} orders")
```

# question 4

```python
import numpy as np

def calculate_revenue(products_matrix):
    # Multiply column 0 (Price) by column 1 (Quantity)
    revenues = products_matrix[:, 0] * products_matrix[:, 1]
    
    print("--- Total Revenue Per Product ---")
    for index, total in enumerate(revenues):
        print(f"Product {index + 1}: ₹{total}")

# Matrix: [Price (₹), Quantity Sold]
flipkart_data = np.array([
    [500, 10],   # Product 1
    [1200, 5],   # Product 2
    [300, 50],   # Product 3
    [1500, 2],   # Product 4
    [800, 15]    # Product 5
])

calculate_revenue(flipkart_data)
```
