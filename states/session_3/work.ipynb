# question 1

```python
import statistics

step_counts = [4500, 7000, 5000, 8000, 4000, 9000, 6000]

step_variance = statistics.variance(step_counts)

print(f"Step Counts: {step_counts}")
print(f"Variance: {step_variance:.1f}")
```

# question 2

```python
import statistics

def calculate_standard_deviation(scores):
    sd = statistics.stdev(scores)
    return round(sd, 2)
exam_scores = [85, 90, 78, 92, 88]
print(f"Exam Scores: {exam_scores}")
print(f"Standard Deviation: {calculate_standard_deviation(exam_scores)}")
```

# question 3

```python
import statistics

friend_a = [200, 200, 200, 200, 200]
friend_b = [100, 300, 150, 400, 50]

sd_a = statistics.stdev(friend_a)
sd_b = statistics.stdev(friend_b)

print(f"Friend A Standard Deviation: {sd_a:.2f}")
print(f"Friend B Standard Deviation: {sd_b:.2f}")

if sd_a < sd_b:
    print("Conclusion: Friend A is more consistent because their standard deviation is lower.")
else:
    print("Conclusion: Friend B is more consistent because their standard deviation is lower.")
```

# question 4

```python
import statistics

salaries = [28000, 29000, 27000, 60000, 26500, 27500]

salary_variance = statistics.variance(salaries)
salary_sd = statistics.stdev(salaries)

print(f"Monthly Salaries: {salaries}")
print(f"Variance: {salary_variance:.2f}")
print(f"Standard Deviation: {salary_sd:.2f}")
```
