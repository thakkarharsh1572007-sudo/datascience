# question 4

```python
import statistics

channel_a = [100, 110, 105, 120, 105]
channel_b = [80, 85, 90, 85, 80]
channel_c = [100, 120, 110, 105, 5000] 

print("--- Channel A ---")
print(f"Mean: {statistics.mean(channel_a):.1f} | Median: {statistics.median(channel_a)} | Mode: {statistics.mode(channel_a)}")

print("\n--- Channel B ---")
print(f"Mean: {statistics.mean(channel_b):.1f} | Median: {statistics.median(channel_b)} | Mode: {statistics.mode(channel_b)}")

print("\n--- Channel C (With Outlier) ---")
print(f"Mean: {statistics.mean(channel_c):.1f} | Median: {statistics.median(channel_c)} | Mode: {statistics.mode(channel_c)}")

print("\nConclusion:")
print("Channel C is highly affected by the 5000-view outlier. The mean jumps to 1087, but the median stays stable at 110, showing the true typical performance.")
```

# question 5

```python
import pandas as pd
import io

csv_data = """Restaurant_Name,Rating,Cost_For_Two
Spicy Grill,4.2,500
Pizza Corner,3.8,800
Burger Joint,4.2,350
Sushi World,4.5,1200
Taco Fiesta,3.5,400
Vegan Bites,4.2,600
Biryani House,4.0,450
Sweet Tooth,4.5,300"""

df = pd.read_csv(io.StringIO(csv_data))

print("--- Swiggy Dataset 'Rating' Statistics ---")
print(f"Mean: {df['Rating'].mean():.2f}")
print(f"Median: {df['Rating'].median()}")
print(f"Mode: {df['Rating'].mode()[0]}")
```
