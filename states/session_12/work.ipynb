# question 1

```python
import numpy as np

# Matrix A: Day 1 data for 2 posts (Rows: Post 1, Post 2. Columns: Likes, Comments)
matrix_a = np.array([
    [100, 15], 
    [250, 40]
])

# Matrix B: Day 2 data for the same metrics
matrix_b = np.array([
    [120, 20], 
    [300, 50]
])

# Perform matrix multiplication using the @ operator
result_matrix = matrix_a @ matrix_b

print("--- Instagram Matrix Multiplication ---")
print("Matrix A (Day 1):\n", matrix_a)
print("\nMatrix B (Day 2):\n", matrix_b)
print("\nResulting Multiplied Matrix:\n", result_matrix)
```

# question 2

```python
# Define the function
def linear_regression_predict(x, m, c):
    return (m * x) + c

# Quick test: if x=5, slope(m)=2, intercept(c)=3, y should be 13
test_x = 5
test_m = 2
test_c = 3
result = linear_regression_predict(test_x, test_m, test_c)

print("--- Linear Regression Function Test ---")
print(f"Inputs: x={test_x}, m={test_m}, c={test_c}")
print(f"Predicted y: {result}")
```

# question 3

```python
def linear_regression_predict(x, m, c):
    return (m * x) + c

distances = [2, 5, 8] # in km
slope_m = 4           # 4 minutes per km
intercept_c = 10      # 10 minutes base prep time

print("--- Zomato Delivery Time Predictions ---")
print(f"Using equation: y = {slope_m}x + {intercept_c}\n")

for distance in distances:
    predicted_time = linear_regression_predict(distance, slope_m, intercept_c)
    print(f"Distance: {distance}km  -->  Predicted Delivery Time: {predicted_time} minutes")
```
