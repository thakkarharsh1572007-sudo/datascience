# question 1

```python
import matplotlib.pyplot as plt

# Data for 7 days
hours_spent = [1.0, 1.5, 2.0, 2.5, 3.0, 3.5, 4.0]
posts_made = [0, 0, 1, 1, 2, 1, 3]

# Create the scatter plot
plt.scatter(hours_spent, posts_made, color='purple', edgecolor='black')
plt.title("Instagram: Hours Spent vs. Posts Made")
plt.xlabel("Hours Spent")
plt.ylabel("Posts Made")
plt.show()

print("Visual Description:")
print("The scatter plot shows an upward trend. As the number of hours spent on the app increases, the number of posts made generally goes up. This visual pattern represents a positive correlation.")
```

# question 2

```python
import numpy as np

# Monthly order counts for 6 friends
zomato_orders = [5, 8, 3, 10, 6, 12]
swiggy_orders = [4, 7, 2, 9, 5, 11]

# np.corrcoef returns a matrix; we extract the correlation between the two arrays
correlation_matrix = np.corrcoef(zomato_orders, swiggy_orders)
r_value = correlation_matrix[0, 1]

print(f"Correlation Coefficient: {r_value:.2f}")

print("\nExplanation:")
print("The correlation is heavily positive (very close to 1). This means there is a strong direct relationship: friends who order a lot of food on Zomato also tend to order a lot of food on Swiggy.")
```

# question 4

```python
import matplotlib.pyplot as plt

# Generated dataset (10 pairs)
# X-axis: Hours spent listening to Spotify per week
spotify_hours = [5, 8, 12, 15, 20, 25, 30, 35, 40, 45]

# Y-axis: Test marks scored (out of 100)
test_marks = [95, 88, 85, 76, 68, 62, 55, 45, 38, 25]

# Create the scatter plot
plt.scatter(spotify_hours, test_marks, color='red', edgecolor='black')
plt.title("Impact of Spotify Time on Test Marks")
plt.xlabel("Hours Spent on Spotify")
plt.ylabel("Marks Scored")
plt.show()

print("Notice how the dots form a downward slope from top-left to bottom-right, which is the classic shape of a strong negative correlation.")
```
