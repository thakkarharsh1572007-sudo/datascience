# question 2

```python
def bayes_posterior(prior, likelihood, evidence):
    return (likelihood * prior) / evidence

# Prior: P(Premium) - The base probability a restaurant is premium
prior_premium = 0.10

# Likelihood: P(Discount | Premium) - Probability of a discount given it is premium
likelihood_discount_given_premium = 0.80

# Additional info for Evidence
prior_regular = 0.90 # (1 - 0.10)
prob_discount_given_regular = 0.20

# Evidence: P(Discount) - Total probability of seeing a discount anywhere
evidence_discount = (likelihood_discount_given_premium * prior_premium) + (prob_discount_given_regular * prior_regular)

# Calculate Posterior: P(Premium | Discount)
posterior = bayes_posterior(prior_premium, likelihood_discount_given_premium, evidence_discount)

print("--- Swiggy Bayes' Theorem ---")
print(f"Total probability of a discount: {evidence_discount}")
print(f"Probability the restaurant is premium given you saw a discount: {posterior:.2f} (or {posterior * 100:.1f}%)")
```

# question 3

```python
def bayes_posterior(prior, likelihood, evidence):
    return (likelihood * prior) / evidence

# 20 simulated email subjects (14 normal, 6 spam)
emails = [
    ("meeting at 5", "not_spam"), ("free gift inside", "spam"), 
    ("project update", "not_spam"), ("claim your free prize", "spam"),
    ("lunch tomorrow?", "not_spam"), ("free coupons for you", "spam"),
    ("invoice attached", "not_spam"), ("buy now, free shipping", "spam"),
    ("family photos", "not_spam"), ("urgent: account locked", "spam"),
    ("see you later", "not_spam"), ("win a free car", "spam"),
    ("weekly report", "not_spam"), ("hello there", "not_spam"),
    ("are we still on?", "not_spam"), ("dental appointment", "not_spam"),
    ("gym tonight?", "not_spam"), ("tax documents", "not_spam"),
    ("happy birthday", "not_spam"), ("don't forget the free snacks", "not_spam")
]

# Calculate basic probabilities from our dataset
total_emails = len(emails)
spam_count = sum(1 for _, label in emails if label == "spam")
free_in_spam = sum(1 for subject, label in emails if "free" in subject and label == "spam")
free_in_all = sum(1 for subject, _ in emails if "free" in subject)

# Prior: P(Spam)
prior_spam = spam_count / total_emails

# Likelihood: P('free' | Spam)
likelihood_free_given_spam = free_in_spam / spam_count

# Evidence: P('free' anywhere)
evidence_free = free_in_all / total_emails

# Posterior: P(Spam | 'free')
prob_spam_given_free = bayes_posterior(prior_spam, likelihood_free_given_spam, evidence_free)

print("--- Spam Filter Simulation ---")
print(f"Probability an email is spam if it contains the word 'free': {prob_spam_given_free:.2f}")
```

```python
# AI Generated Scenario:
# On Flipkart, 15% of all users complete a purchase. If a user completes a purchase, there is a 90% chance they added the item to their wishlist first. For users who do not complete a purchase, there is still a 20% chance they added the item to their wishlist. What is the probability a user will actually buy an item given that they added it to their wishlist?
def bayes_posterior(prior, likelihood, evidence):
    return (likelihood * prior) / evidence

# Prior: P(Buy) 
prior_buy = 0.15

# Likelihood: P(Wishlist | Buy)
likelihood_wishlist_given_buy = 0.90

# P(Not Buy) and P(Wishlist | Not Buy)
prior_not_buy = 0.85 
prob_wishlist_given_not_buy = 0.20

# Evidence: P(Wishlist)
evidence_wishlist = (likelihood_wishlist_given_buy * prior_buy) + (prob_wishlist_given_not_buy * prior_not_buy)

# Posterior: P(Buy | Wishlist)
posterior_buy_given_wishlist = bayes_posterior(prior_buy, likelihood_wishlist_given_buy, evidence_wishlist)

print("--- Flipkart Wishlist Conversion ---")
print(f"Probability a user buys the item after adding it to their wishlist: {posterior_buy_given_wishlist:.2f} (or {posterior_buy_given_wishlist * 100:.1f}%)")
```
