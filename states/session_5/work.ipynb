```python
print("--- Bias in Spotify Survey ---")
print("Type of Bias: Selection Bias (specifically, Undercoverage or Sampling Bias)")
print("Explanation: By only sending the survey to users who recently streamed Bollywood music, the sample completely excludes fans of other genres like Hip-Hop, Classical, or Rock. The results will be heavily skewed toward Bollywood preferences and will fail to represent the true, diverse music tastes of the entire Spotify user base.")
```

```python
print("--- Sampling Techniques for BookMyShow ---\n")

print("Technique: Stratified Sampling")
print("How to use it: You divide the entire BookMyShow user base into distinct subgroups (strata) based on the city they live in (e.g., Mumbai, Delhi, Bangalore). You then take a random sample of users from *each* city. This ensures that smaller cities are accurately represented in your study, rather than just getting responses from massive metropolitan areas.\n")

print("Technique: Systematic Sampling")
print("How to use it: You generate a sequential list of all users who successfully booked a movie ticket today. To form your sample, you select a random starting point and then pick every 10th user on that list to receive an email survey asking about their checkout experience.")
```
