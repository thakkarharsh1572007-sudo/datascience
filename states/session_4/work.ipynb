# question 1

```python
import matplotlib.pyplot as plt

swiggy_orders = [12, 15, 17, 14, 13, 16, 200, 18, 14, 15]

plt.hist(swiggy_orders, bins=20, color='skyblue', edgecolor='black')
plt.title("Swiggy Orders (Right Skewed)")
plt.xlabel("Number of Orders")
plt.ylabel("Frequency")
plt.show()

print("Shape: Right-skewed (Positively skewed).")
print("Reasoning: The extreme outlier of 200 orders creates a long tail stretching to the right, while the typical days cluster heavily on the left.")
```

# question 2

```python
import statistics

followers_thousands = [5, 7, 8, 8, 9, 10, 12, 15, 95]

mean_val = statistics.mean(followers_thousands)
median_val = statistics.median(followers_thousands)

print(f"Mean: {mean_val:.2f}")
print(f"Median: {median_val}")

if mean_val > median_val:
    print("\nConclusion: The Mean is significantly higher than the Median. This tells us the single 95k account is pulling the average up, making the data Right-Skewed.")
```

# question 3

```python
import matplotlib.pyplot as plt

typical_views = [100, 120, 110, 150, 130, 90, 140, 105, 115, 125]
viral_views = [5000, 8000]
youtube_data = typical_views + viral_views

plt.hist(youtube_data, bins=15, color='salmon', edgecolor='black')
plt.title("YouTube Video Views (Right Skewed)")
plt.xlabel("Views")
plt.ylabel("Number of Videos")
plt.show()

print("Dataset: YouTube Video Views")
print("Why it is Right-Skewed: Most videos get a standard, lower amount of views (creating a peak on the left). However, rare viral hits get massive numbers, stretching the tail infinitely to the right.")
```

# question 4

```python
import matplotlib.pyplot as plt

flipkart_ratings = [3, 3, 4, 4, 4, 5, 5, 5, 5, 5]

plt.hist(flipkart_ratings, bins=[0.5, 1.5, 2.5, 3.5, 4.5, 5.5], color='lightgreen', edgecolor='black')
plt.title("Flipkart Ratings (Left Skewed)")
plt.xlabel("Star Rating")
plt.ylabel("Frequency")
plt.show()

print("Shape: Left-skewed (Negatively skewed).")
print("Justification: Most users leave a 5-star rating (creating a large peak on the right), with the tail extending to the left toward the less frequent lower ratings.")
```
