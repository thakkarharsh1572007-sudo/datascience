# question 2

```python
from scipy.stats import ttest_ind_from_stats

mean_a, std_a, n_a = 350, 50, 30

mean_b, std_b, n_b = 320, 50, 30

t_stat, p_value = ttest_ind_from_stats(mean1=mean_a, std1=std_a, nobs1=n_a,
                                       mean2=mean_b, std2=std_b, nobs2=n_b)

alpha = 0.05

print("--- Zomato A/B Test Results ---")
print(f"P-Value: {p_value:.4f}")

if p_value < alpha:
    print("Conclusion: Reject the Null Hypothesis (H0).")
    print("Explanation: The new home page statistically significantly increased the average order value.")
else:
    print("Conclusion: Fail to reject the Null Hypothesis (H0).")
    print("Explanation: The difference in order value is not statistically significant and could just be due to chance.")
```
