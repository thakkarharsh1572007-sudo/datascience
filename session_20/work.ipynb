```python
class Flipkart_style_app :
    def __init__(self):
        self.__product_price = 0
    def set_price(self,price):
        self.__product_price = price 
    def get_price(self):
        return self.__product_price
product = Flipkart_style_app()
product.set_price(200)
print(product.get_price())
```

```python
from abc import ABC, abstractmethod

class PaymentMethod(ABC):
    @abstractmethod
    def pay(self, amount):
        pass

class Paytm(PaymentMethod):
    def pay(self, amount):
        if amount <= 0:
            print("Please recall this method with a positive amount, sorry for the inconvenience! .....")
        else:
            print("Thank you for connecting with PayTM! \nPayment of Rs.", amount, "is done.")

class PhonePay(PaymentMethod):
    def pay(self, amount):
        if amount <= 0:
            print("Please recall this method with a positive amount, sorry for the inconvenience! .....")
        else:
            print(f"Payment of Rs. {amount} is successful.")
            print("Thank you for using PhonePay!.......") 

b = Paytm()
c = PhonePay()

b.pay(100)
print("")
c.pay(100)
```
