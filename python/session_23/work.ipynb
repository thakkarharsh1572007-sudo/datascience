# question 2

```python
import datetime
import requests

url = "https://api.coingecko.com/api/v3/simple/price"
params = {
    "ids": "bitcoin,ethereum",
    "vs_currencies": "usd,inr"
}

response = requests.get(url, params=params)
data = response.json()
now = datetime.datetime.now().strftime("%Y-%m-%d %H:%M:%S")

print(f"=== Crypto Prices as of {now} ===")
print(f"Bitcoin (BTC)  : ${data['bitcoin']['usd']:,.2f} USD | ₹{data['bitcoin']['inr']:,.2f} INR")
print(f"Ethereum (ETH) : ${data['ethereum']['usd']:,.2f} USD | ₹{data['ethereum']['inr']:,.2f} INR")
```

# question 3

```python
import requests

API_KEY = "DEMO_KEY"
url = f"https://api.nasa.gov/planetary/apod?api_key={API_KEY}"

response = requests.get(url)
data = response.json()

print(f" NASA APOD: {data.get('title')} ({data.get('date')})")
print("-" * 50)
print(f"Explanation:\n{data.get('explanation')}\n")

# Get image URL and download it
img_url = data.get("hdurl") or data.get("url")
print(f"Downloading image from: {img_url}")

img_response = requests.get(img_url)
with open("nasa_apod_today.jpg", "wb") as f:
    f.write(img_response.content)

print("Image successfully saved")
```

# question 4

```python
import requests

url = "https://disease.sh/v3/covid-19/countries/india"

response = requests.get(url)
data = response.json()

total_cases = f"{data.get('cases', 0):,}"
recovered = f"{data.get('recovered', 0):,}"
deaths = f"{data.get('deaths', 0):,}"
active = f"{data.get('active', 0):,}"

divider = f"+{'-' * 22}+{'-' * 20}+"

print("\n COVID-19 Statistics: India")
print(divider)
print(f"| {'Metric':<20} | {'Count':>18} |")
print(divider)
print(f"| {'Total Cases':<20} | {total_cases:>18} |")
print(f"| {'Recovered':<20} | {recovered:>18} |")
print(f"| {'Deaths':<20} | {deaths:>18} |")
print(f"| {'Active Cases':<20} | {active:>18} |")
print(divider)
```

```python
import requests

OPENWEATHER_API_KEY = "63450de986dac6821f097afc0da0b78e"
WEATHER_URL = f"https://api.openweathermap.org/data/2.5/weather?q=Mumbai&appid={OPENWEATHER_API_KEY}&units=metric"
CRYPTO_URL = "https://api.coingecko.com/api/v3/simple/price?ids=bitcoin&vs_currencies=usd"

print(" Fetching combined data...\n")

# Fetch Weather Data
w_res = requests.get(WEATHER_URL)
w_data = w_res.json()
weather_info = f"{w_data['main']['temp']}°C ({w_data['weather'][0]['description'].title()})"

# Fetch Bitcoin Data
c_res = requests.get(CRYPTO_URL)
c_data = c_res.json()
btc_info = f"${c_data['bitcoin']['usd']:,.2f} USD"

# Output Dashboard
print("=" * 45)
print("         LIVE DASHBOARD REPORT")
print("=" * 45)
print(f" Mumbai Weather : {weather_info}")
print(f" Bitcoin Price  : {btc_info}")
print("=" * 45)
```
