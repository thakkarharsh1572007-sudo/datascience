# question 1

```python
import requests
import json

url = "https://api.binance.com/api/v3/ticker/24hr"
symbols = '["BTCUSDT","ETHUSDT","BNBUSDT","SOLUSDT","XRPUSDT","ADAUSDT","DOGEUSDT","LTCUSDT","LINKUSDT","TRXUSDT"]'

response = requests.get(url, params={'symbols': symbols})
data = response.json()

with open("crypto_data.json", "w") as file:
    json.dump(data, file, indent=4)

print("Data successfully saved to crypto_data.json")
```

# question 2

```python
import json

def find_most_volatile_coin(data):
    most_volatile = max(data, key=lambda x: abs(float(x['priceChangePercent'])))
    return most_volatile['symbol']

with open("crypto_data.json", "r") as file:
    crypto_data = json.load(file)

volatile_symbol = find_most_volatile_coin(crypto_data)
print(f"The most volatile coin in the last 24 hours is: {volatile_symbol}")
```

# question 3

```python
import json

with open("crypto_data.json", "r") as file:
    crypto_data = json.load(file)

total_price = sum(float(coin['lastPrice']) for coin in crypto_data)
average_price = total_price / len(crypto_data)

print(f"Market Average Price: ${average_price:,.4f}\n")
print("Coins trading below the average price:")

for coin in crypto_data:
    price = float(coin['lastPrice'])
    if price < average_price:
        print(f"{coin['symbol']}: ${price:,.4f}")
```

# question 4

```python
import json

def rank_coins_by_volume(data):
    sorted_coins = sorted(data, key=lambda x: float(x['quoteVolume']), reverse=True)
    
    print("Top 5 Coins by Traded Volume (24h):")
    for index in range(5):
        coin = sorted_coins[index]
        volume = float(coin['quoteVolume'])
        print(f"{index + 1}. {coin['symbol']} - Volume: ${volume:,.2f}")

with open("crypto_data.json", "r") as file:
    crypto_data = json.load(file)

rank_coins_by_volume(crypto_data)
```

# question 5

```python
import time
import requests
import schedule


def fetch_data_with_retry():
  url = "https://api.binance.com/api/v3/ticker/24hr"
  symbols = '["BTCUSDT","ETHUSDT","BNBUSDT","SOLUSDT","XRPUSDT","ADAUSDT","DOGEUSDT","LTCUSDT","LINKUSDT","TRXUSDT"]'
  backoff = 2

  while True:
    response = requests.get(url, params={"symbols": symbols})
    if response.status_code == 429:
      print(f"Rate limit hit! Retrying in {backoff} seconds...")
      time.sleep(backoff)
      backoff *= 2
    elif response.status_code == 200:
      print(f"Success! Data fetched at {time.strftime('%H:%M:%S')}")
      break
    else:
      print(f"Failed with HTTP Error: {response.status_code}")
      break


schedule.clear()  
schedule.every(10).seconds.do(fetch_data_with_retry)

print("Starting automation. Running first fetch now...")
fetch_data_with_retry()

try:
  while True:
    schedule.run_pending()
    time.sleep(1)
except KeyboardInterrupt:
  print("\nScheduler stopped manually by user.")
```
