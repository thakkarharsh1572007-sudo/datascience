# Question 1

```python
import requests

url = 'https://jsonplaceholder.typicode.com/posts'
response = requests.get(url)

print(f"Status Code: {response.status_code}")

posts = response.json()
if posts:
    print(f"First post title: {posts[0]['title']}")
```

# Question 3

```python
import requests

url = 'https://jsonplaceholder.typicode.com/users'
response = requests.get(url)
users = response.json()

print("Usernames with '.org' emails:")
for user in users:
    if user['email'].endswith('.org'):
        print(user['username'])
```

# Question 4

```python
import requests

url = 'http://www.omdbapi.com/'
query_params = {
    'apikey': 'demo',
    's': 'Avengers'
}

response = requests.get(url, params=query_params)
data = response.json()

total_results = data.get('totalResults', '0')
print(f"Total results found for 'Avengers': {total_results}")
```
