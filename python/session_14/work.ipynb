# question 3

```python
import csv
match_data = [
    ["match_id", "team1","team2","winner"],
    ["01","csk","rcb","csk"],
    ["02","csk","gt","gt"],
    ["03","gt","rcb","gt"]
]
file = open ("ipldata.csv","w",newline='')
writter = csv.writer(file)
writter.writerows(match_data)
file.close()
```

```python
file = open("ipldata.csv","r")
a= file.readlines()
file.close()
a
```

# question 4

```python
import json
trending_movies_input = [
    {"title":"hfhfh","year":"2025","reating": "9.8/10"},
    {"title":"hfhfh2","year":"2025","reating": "9.6/10"},
    {"title":"hfhfh3","year":"2026","reating": "9.3/10"}
]
file = open("trending_movie.json","w")

json.dump(trending_movies_input,file,indent=False)
file.close()
```

```python
file = open("trending_movie.json","r")
a = json.load(file)
file.close()
a
```
