# Question 1

```python
class Product :
    def __init__ (self,name = "",price = 0):
        self.name = name 
        self.price = price 
    def get_disconted_price (self):
        return (self.price-(self.price/10))
a = Product("apple",100)
print (a.get_disconted_price())
```

# Question 2

```python
class Electronics (Product):
    def __init__(self,name = "",price = 0):
        super().__init__(name,price)
    def get_dicounted_price (self):
        return (self.price-(self.price/20))
b = Electronics("cpu",250000)
print(b.get_dicounted_price())
```
