```python
import requests

api_key = "63450de986dac6821f097afc0da0b78e"
city = "Ahmedabad"

url = "https://api.openweathermap.org/data/2.5/weather"

params = {
    "q": city,
    "appid": api_key,
    "units": "metric" 
}

response = requests.get(url, params=params)

if response.status_code == 200:
    data = response.json()
    temp = data["main"]["temp"]
    print(f"The current temperature in {city} is {temp}°C.")
else:
    print(f"Error: {response.status_code} - {response.text}")
```

# question 3

```python
import asyncio
import httpx

async def fetch_spotify_trending(client):
    url = "https://jsonplaceholder.typicode.com/posts/1"
    response = await client.get(url)
    return f"Spotify Data: {response.json().get('title')}"

async def fetch_bookmyshow_trending(client):
    url = "https://jsonplaceholder.typicode.com/posts/2"
    response = await client.get(url)
    return f"BookMyShow Data: {response.json().get('title')}"

async def main():
    # Use an AsyncClient to manage connections
    async with httpx.AsyncClient() as client:
        # Schedule both tasks to run concurrently
        results = await asyncio.gather(
            fetch_spotify_trending(client),
            fetch_bookmyshow_trending(client)
        )
        
        for result in results:
            print(result)

await main()
```

```python
import urllib.parse
import secrets

def generate_github_login_url():
    # Replace with your actual GitHub Client ID
    client_id = "Ov23liDzlvud8ejsdvN7" 
    
    # This matches what you saved in GitHub exactly
    redirect_uri = "http://localhost:8888/callback"
    
    # GitHub's authorization endpoint
    auth_base_url = "https://github.com/login/oauth/authorize"
    
    # The permissions your app is requesting (e.g., reading user profile)
    scope = "read:user"
    
    # Security token to prevent CSRF attacks
    state = secrets.token_hex(16)
    
    # Construct the query parameters
    params = {
        "client_id": client_id,
        "redirect_uri": redirect_uri,
        "scope": scope,
        "state": state
    }
    
    # Encode the parameters safely into a URL string
    url_args = urllib.parse.urlencode(params)
    login_url = f"{auth_base_url}?{url_args}"
    
    return login_url

print("Redirect the user to this URL to log in to GitHub:")
print(generate_github_login_url())
```
