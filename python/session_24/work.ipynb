```python
import requests

url = "https://api.coingecko.com/api/v3/coins/markets"
params = {
    "vs_currency": "usd",
    "order": "market_cap_desc",
    "per_page": 10,
    "page": 1
}

response = requests.get(url, params=params)
coins = response.json()

print("Top 10 Cryptocurrencies by Market Cap:")
print("-" * 40)
for coin in coins:
    name = coin["name"]
    price = coin["current_price"]
    print(f"{name}: ${price:,.2f}")
```

# question 3

```python
import csv
import requests

url = "https://api.coingecko.com/api/v3/coins/markets"
params = {
    "vs_currency": "usd",
    "order": "market_cap_desc",
    "per_page": 10,
    "page": 1
}

response = requests.get(url, params=params)
coins = response.json()

filename = "crypto_prices.csv"

# Write data to CSV
with open(filename, mode="w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)
    
    # Write header row
    writer.writerow(["Name", "Current Price (USD)", "24h Change (%)", "24h High (USD)", "24h Low (USD)"])
    
    # Write data rows
    for coin in coins:
        writer.writerow([
            coin.get("name"),
            coin.get("current_price"),
            coin.get("price_change_percentage_24h"),
            coin.get("high_24h"),
            coin.get("low_24h")
        ])

print(f"Data successfully saved to '{filename}'.")
```

# question 4

```python
import csv
import requests

URL = "https://api.coingecko.com/api/v3/coins/markets"
PARAMS = {
    "vs_currency": "usd",
    "order": "market_cap_desc",
    "per_page": 10,
    "page": 1
}
HEADERS = {
    "User-Agent": "CryptoDataCollector/1.0"
}
CSV_FILE = "crypto_prices.csv"


def fetch_crypto_data():
    try:
        response = requests.get(URL, params=PARAMS, headers=HEADERS, timeout=10)
        
        # Check HTTP status codes
        if response.status_code == 429:
            print("Error: Rate limit exceeded. CoinGecko allows limited requests per minute for public access. Please wait and try again.")
            return None
        
        response.raise_for_status()
        return response.json()

    except requests.exceptions.ConnectionError:
        print("Network Error: Could not connect to CoinGecko. Please check your internet connection.")
    except requests.exceptions.Timeout:
        print("Timeout Error: The request took too long to respond.")
    except requests.exceptions.HTTPError as http_err:
        print(f"HTTP Error occurred: {http_err} (Status Code: {response.status_code})")
    except requests.exceptions.RequestException as err:
        print(f"An unexpected request error occurred: {err}")
    except ValueError:
        print("Data Error: Failed to parse JSON response from the server.")
    
    return None


def display_and_save(coins):
    if not coins or not isinstance(coins, list):
        print("No valid coin data received.")
        return

    # 1. Print formatted table to console
    print(f"\n{'Name':<18} {'Price (USD)':<14} {'24h Change (%)':<16} {'24h High':<14} {'24h Low':<14}")
    print("-" * 76)

    rows_to_save = []
    for coin in coins:
        name = coin.get("name", "N/A")
        price = coin.get("current_price") or 0.0
        change_24h = coin.get("price_change_percentage_24h") or 0.0
        high_24h = coin.get("high_24h") or 0.0
        low_24h = coin.get("low_24h") or 0.0

        print(f"{name:<18} ${price:<13,.2f} {change_24h:>+13.2f}% ${high_24h:<13,.2f} ${low_24h:<13,.2f}")

        rows_to_save.append([name, price, change_24h, high_24h, low_24h])

    # 2. Save data to CSV file
    try:
        with open(CSV_FILE, mode="w", newline="", encoding="utf-8") as file:
            writer = csv.writer(file)
            writer.writerow(["Name", "Current Price (USD)", "24h Change (%)", "24h High (USD)", "24h Low (USD)"])
            writer.writerows(rows_to_save)
        print(f"\n Data successfully saved to '{CSV_FILE}'.")
    except OSError as err:
        print(f"\nFile Error: Could not write to '{CSV_FILE}': {err}")


if __name__ == "__main__":
    print("Fetching top 10 cryptocurrencies from CoinGecko...")
    crypto_data = fetch_crypto_data()
    if crypto_data:
        display_and_save(crypto_data)
```
