```python
from functools import reduce
try:
    def get_song_duration_per_minute(playlist):
        n = len(playlist)
        duration = reduce(lambda x,y:x+y,playlist)
        return duration/n
except Exception as e:
    print ("something got wrong !")
finally :
    print("function call is ended")
playlist = [12,2,6,35,0,25,24,35,26]
print("song duration per mints in this plalist is : ",get_song_duration_per_minute(playlist))
```
