# question 1

```python
import requests
url = "https://jsonplaceholder.typicode.com/posts"
payload = {
    "title": "My New Post",
    "body": "This is the content of the post.",
    "userId": 1
}
response = requests.post(url, json=payload)
print(f"Status Code: {response.status_code}")
print("Response JSON:")
print(response.json())
```

# question 2

```python
import requests
playlist_name = input("Enter a new playlist name: ")
playlist_desc = input("Enter playlist description: ")
url = "https://jsonplaceholder.typicode.com/posts"
payload = {
    "name": playlist_name,
    "description": playlist_desc
}
response = requests.post(url, json=payload)
data = response.json()

print(f"\nSuccess! Your playlist was created with ID: {data.get('id')}")
```

# question 4

```python
import requests
import csv

url = "https://jsonplaceholder.typicode.com/posts"
response = requests.get(url)
posts = response.json()

latest_posts = posts[:5]

with open('posts.csv', 'w', newline='', encoding='utf-8') as csvfile:
    writer = csv.writer(csvfile)
    writer.writerow(['userId', 'title'])
    for post in latest_posts:
        writer.writerow([post['userId'], post['title']])

print("Successfully saved 5 posts to posts.csv")
```

# question 5

```python
import requests
import json

url = "https://jsonplaceholder.typicode.com/posts"
response = requests.get(url)
posts = response.json()

latest_posts = posts[:5]

with open('posts.json', 'w', encoding='utf-8') as jsonfile:
    json.dump(latest_posts, jsonfile, indent=4)

print("Successfully saved 5 posts to posts.json")
```
