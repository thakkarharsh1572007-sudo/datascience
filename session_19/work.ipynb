```python
class FoodOrder :
    def __init__(self,restaurant_name,items,total_price):
        self.restaurant_name = restaurant_name
        self.items = items
        self.total_amount = total_price
    def getDeaitals (self):
        print(f"you order from {self.restaurant_name} \n items are {self.items}\n  and your total amount is {self.total_amount}")
    def addItem(self,item_name ,item_price):
        self.items.append(item_name)
        self.total_amount+=item_price
        print ("item was added successfully !.........")
a = FoodOrder("point restaurent",["pizza","water bottel","papsi"],250)
a.addItem("browne",100)
a.getDeaitals()
```
