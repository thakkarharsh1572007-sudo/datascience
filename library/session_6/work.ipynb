# question 4

```python
import pandas as pd
import numpy as np
data = {
    "restaurant_name": ["Agashiye", "Upper Crust", "Real Lassi", "Gwalia Sweets", "Rp's Pizza"],
    "cuisine": ["Gujarati", "Bakery", "Beverages", "Fast Food", "Italian"],
    "rating": [np.nan, 4.3, np.nan, 4.1, 4.5]  
}
df = pd.DataFrame(data)
print("--- Before Filling Missing Ratings ---")
print(df)
print("\n" + "="*50 + "\n")
df['rating'] = df['rating'].ffill()
df['rating'] = df['rating'].bfill()
print("--- After Forward Fill and Backward Fill ---")
print(df)
```
