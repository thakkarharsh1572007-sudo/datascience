```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

delivery_times = np.random.normal(loc=30, scale=8, size=50)

plt.figure(figsize=(8, 5))
sns.histplot(delivery_times, bins=10, color='tomato', kde=True)
plt.title('Zomato Delivery Times (50 Orders)')
plt.xlabel('Delivery Time (Minutes)')
plt.ylabel('Frequency')
```

# question 2

```python
tips = sns.load_dataset('tips')

plt.figure(figsize=(8, 5))
sns.boxplot(data=tips, x='day', y='total_bill', hue='day', palette='Set2', legend=False)
plt.title('Total Bill Amount by Day of the Week')
plt.xlabel('Day')
plt.ylabel('Total Bill ($)')
```

# question 3

```python
sns.set_theme(style='darkgrid')

teams = ['CSK', 'MI', 'RCB', 'KKR', 'DC', 'PBKS', 'SRH', 'RR']
simulated_teams = np.random.choice(teams, size=160) 
simulated_scores = np.random.normal(loc=165, scale=25, size=160)

plt.figure(figsize=(10, 6))
sns.violinplot(x=simulated_teams, y=simulated_scores, hue=simulated_teams, palette='muted', legend=False)
plt.title('IPL Run Distributions by Team')
plt.xlabel('Team')
plt.ylabel('Runs Scored')
```

# question 4

```python
genres = ['Pop', 'Hip-Hop', 'Indie', 'Electronic']
track_list = np.random.choice(genres, size=40, p=[0.4, 0.3, 0.2, 0.1])

plt.figure(figsize=(8, 5))
sns.countplot(x=track_list, hue=track_list, palette='viridis', legend=False)
plt.title('Spotify Playlist: Tracks per Genre')
plt.xlabel('Genre')
plt.ylabel('Number of Songs')
```

```python
sns.set_theme(style='whitegrid')

step_counts = [5500, 8200, 10500, 7800, 6100, 12000, 9400]

plt.figure(figsize=(8, 5))
sns.kdeplot(step_counts, color='mediumseagreen', fill=True, linewidth=2)
plt.title('Density of Daily Step Counts (Past Week)')
plt.xlabel('Steps')
plt.ylabel('Density')
```
