```python
import pandas as pd
import sweetviz as sv
df_zomato = pd.read_csv('zomato_dataset.csv')
df_zomato.columns = df_zomato.columns.str.strip()
df_mumbai = df_zomato[df_zomato['City'].astype(str).str.strip() == 'Mumbai']
df_delhi = df_zomato[df_zomato['City'].astype(str).str.strip() == 'New Delhi'] # Note: It might be 'Delhi' or 'New Delhi' in your dataset
print(f"Found {len(df_mumbai)} rows for Mumbai and {len(df_delhi)} rows for Delhi.")
comparison_report = sv.compare(
    [df_mumbai, "Mumbai Restaurants"], 
    [df_delhi, "Delhi Restaurants"]
)
comparison_report.show_html('zomato_comparison.html')
```

# question 4

```python
import pandas as pd
import re
df_flipkart = pd.read_csv('Dataset-SA.csv')
df_flipkart = df_flipkart.dropna(subset=['Summary'])
def clean_text(text):
    text = str(text).lower()
    text = re.sub(r'[^\w\s]', '', text) 
    return text
df_flipkart['cleaned_summary'] = df_flipkart['Summary'].apply(clean_text)
df_flipkart['Rate'] = pd.to_numeric(df_flipkart['Rate'], errors='coerce')
df_flipkart['Rate'] = df_flipkart['Rate'].fillna(df_flipkart['Rate'].median())
print(df_flipkart[['Summary', 'cleaned_summary', 'Rate']].head())
```

```python
import pandas as pd
from ydata_profiling import ProfileReport

def generate_swiggy_report(file_path):
    print("Loading Swiggy dataset...")
    df_swiggy = pd.read_csv(file_path)
    print(f"Dataset loaded successfully with {df_swiggy.shape[0]} rows and {df_swiggy.shape[1]} columns.")
    print("Generating comprehensive EDA report (this may take a moment)...")
    swiggy_profile = ProfileReport(
        df_swiggy,
        title="Swiggy Food Orders - Automated EDA",
        explorative=True
    )
    output_file = "swiggy_summary_report.html"
    swiggy_profile.to_file(output_file)
    print(f"Success! Open '{output_file}' in your web browser to view the report.")
generate_swiggy_report('swiggy_orders.csv')
```
