# question 4

```python
np.set_printoptions(formatter={'float': '{:0.2f}'.format})
rating = np.random.uniform(3,5,(3,3))
rating
```

```python
np.linalg.inv(rating)
```

```python
np.linalg.det(rating)
```

```python
eigval,eigvactor = np.linalg.eig(rating)
print("eigen values : \n",eigval,"\n\n\neigen vactors : \n",eigvactor)
```
