# question 3

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

data = {
    'Platform': ['Swiggy', 'Zomato', "Domino's"],
    '1': [10, 4, 1],
    '2': [5, 6, 3]
}
df = pd.DataFrame(data)
fig, ax = plt.subplots(figsize=(8, 6))
x = np.arange(len(df['Platform']))
width = 0.35
ax.bar(x - width/2, df['1'], width, label='Person 1')
ax.bar(x + width/2, df['2'], width, label='Person 2')
ax.set_xlabel('Delivery Platform')
ax.set_ylabel('Number of Orders (Last Month)')
ax.set_title('Food Delivery Orders by Platform (2 People)')
ax.set_xticks(x)
ax.set_xticklabels(df['Platform'])
ax.legend()
plt.show()
```

# question 5

```python
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
screen_time_mins = [45, 50, 35, 60, 85, 120, 90]
posts_liked = [12, 15, 8, 25, 40, 55, 30]
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))
ax1.plot(days, screen_time_mins, marker='o', color='magenta', linestyle='-', linewidth=2)
ax1.set_title('Daily Instagram Screen Time')
ax1.set_xlabel('Day of the Week')
ax1.set_ylabel('Screen Time (Minutes)')
ax1.grid(True, linestyle='--', alpha=0.6)
ax2.bar(days, posts_liked, color='cornflowerblue', edgecolor='black')
ax2.set_title('Instagram Posts Liked per Day')
ax2.set_xlabel('Day of the Week')
ax2.set_ylabel('Number of Likes')
ax2.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.savefig('social_media_usage.png', dpi=300, bbox_inches='tight')
```
