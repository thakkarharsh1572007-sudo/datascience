# question 2

```python
import seaborn as sns
import matplotlib.pyplot as plt

flipkart_df = pd.read_csv('flipkart_10k_reviews.csv')
plt.figure(figsize=(8, 5))
sns.countplot(data=flipkart_df, x='rating', palette='viridis')
plt.title('Flipkart: Review Counts per Rating')
plt.xlabel('Rating (1-5 Stars)')
plt.ylabel('Count')
plt.show()
```

# question 3

```python
zomato_df = pd.read_csv('zomato_dataset.csv')
plt.figure(figsize=(8, 6))
sns.scatterplot(data=zomato_df, x='average_cost_for_two', y='user_rating', color='coral', alpha=0.7)
plt.title('Zomato: Average Cost for Two vs. User Rating')
plt.xlabel('Average Cost for Two')
plt.ylabel('User Rating')
```

# question 4

```python
bms_df = pd.read_csv('bookmyshow_dataset.csv')
genre_avg = bms_df.groupby('genre')['box_office_collection'].mean().reset_index()
sorted_genre_avg = genre_avg.sort_values(by='box_office_collection', ascending=False)
print(sorted_genre_avg.to_string(index=False))
```
