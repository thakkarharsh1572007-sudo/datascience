# question 1

```python
import pandas as pd
from sqlalchemy import create_engine, text

USER = "root"
PASSWORD = "123456"
HOST = "localhost"
DB_NAME = "entertainment_db"

server_engine = create_engine(f"mysql+pymysql://{USER}:{PASSWORD}@{HOST}/")

with server_engine.connect().execution_options(isolation_level="AUTOCOMMIT") as conn:
    conn.execute(text(f"CREATE DATABASE IF NOT EXISTS {DB_NAME}"))

db_uri = f"mysql+pymysql://{USER}:{PASSWORD}@{HOST}/{DB_NAME}"
engine = create_engine(db_uri)
restaurants_data = {
    'id': [1, 2, 3, 4, 5, 6],
    'name': ['Spicy Venue', 'Burger King', 'Truffles', 'Paradise Biryani', 'Dominos', 'Subway'],
    'cuisine': ['Andhra', 'Fast Food', 'Continental', 'Biryani', 'Pizza', 'Healthy'],
    'rating': [4.5, 3.8, 4.7, 4.2, 4.0, 3.9]
}
df_dummy_restaurants = pd.DataFrame(restaurants_data)

movies_data = {
    'id': [1, 2, 3, 4, 5],
    'name': ['Inception', 'The Room', 'Interstellar', 'Cats', 'The Dark Knight'],
    'rating': [8.8, 3.6, 8.6, 2.8, 9.0],
    'genre': ['Sci-Fi', 'Drama', 'Sci-Fi', 'Musical', 'Action']
}
df_dummy_movies = pd.DataFrame(movies_data)

df_dummy_restaurants.to_sql('restaurants', con=engine, if_exists='replace', index=False)
df_dummy_movies.to_sql('movies', con=engine, if_exists='replace', index=False)

print("Database and tables populated successfully!\n")
```

```python
df_restaurants = pd.read_sql('restaurants', con=engine)
print("First 5 rows of the 'restaurants' table:")
print(df_restaurants.head())
```

# question 2

```python
query = "SELECT name, rating FROM movies WHERE rating > 8"
df_movies = pd.read_sql_query(query, con=engine)
print("Filtered 'movies' table (Rating > 8):")
print(df_movies)
```

# question 4

```python
users_data = {
    "user_id": [1, 2, 3],
    "username": ["alice_smith", "bob_jones", "charlie_brown"]
}
users_df = pd.DataFrame(users_data)
users_df.to_csv("users.csv", index=False)

orders_data = {
    "order_id": [101, 102, 103],
    "user_id": [1, 2, 1],
    "amount": [250.00, 150.50, 99.99]
}
orders_df = pd.DataFrame(orders_data)
orders_df.to_csv("orders.csv", index=False)

print("Sample CSV files created successfully!")
```

```python
from pathlib import Path
df1 = pd.read_csv(Path("orders.csv"))
df2 = pd.read_csv(Path("users.csv"))
pd.merge(df2,df1,on="user_id")[["username","amount"]]
```
