# question 1

```python
import matplotlib.pyplot as plt
import numpy as np

fig, axs = plt.subplots(2, 2, figsize=(10, 8))

x = [1, 2, 3, 4, 5]
y_line = [2, 4, 6, 8, 10]
axs[0, 0].plot(x, y_line, color='blue')
axs[0, 0].set_title('Line Plot')

y_bar = [10, 20, 15, 25, 30]
axs[0, 1].bar(x, y_bar, color='green')
axs[0, 1].set_title('Bar Chart')

x_scatter = np.random.rand(50)
y_scatter = np.random.rand(50)
axs[1, 0].scatter(x_scatter, y_scatter, color='red')
axs[1, 0].set_title('Scatter Plot')

sizes = [20, 30, 40, 10]
labels = ['A', 'B', 'C', 'D']
axs[1, 1].pie(sizes, labels=labels, autopct='%1.1f%%')
axs[1, 1].set_title('Pie Chart')

plt.tight_layout()
```

# question 2

```python
platforms = ['Zomato', 'Swiggy', "Domino's"]
times = [35, 42, 25]
colors = ['#E23744', '#FC8019', '#0055A5']
linestyles = ['--', '-.', ':']

fig, ax = plt.subplots(figsize=(8, 6))

for i in range(len(platforms)):
    ax.bar(platforms[i], times[i], color=colors[i], edgecolor='black', linewidth=2.5, linestyle=linestyles[i])

ax.set_xlabel('Platform')
ax.set_ylabel('Average Delivery Time (mins)')
ax.set_title('Food Delivery Times')
```

# question 3

```python
influencers = ['Person A', 'Person B', 'Person C', 'Person D', 'Person E']
followers = [15000, 32000, 8000, 45000, 21000]
posts = [1, 4, 2, 5, 3]

fig, ax1 = plt.subplots(figsize=(10, 6))

ax1.bar(influencers, followers, color='cyan', alpha=0.6)
ax1.set_xlabel('Influencers')
ax1.set_ylabel('Instagram Followers', color='blue')
ax1.tick_params(axis='y', labelcolor='blue')

ax2 = ax1.twinx()
ax2.plot(influencers, posts, color='magenta', marker='D', linewidth=2)
ax2.set_ylabel('Average Daily Posts', color='magenta')
ax2.tick_params(axis='y', labelcolor='magenta')

plt.title('Influencer Stats: Followers vs Daily Posts')
```

# question 4

```python
movies = ['Pathaan', 'Jawan', 'Animal', 'Gadar 2', 'Dangal']
ratings = [6.5, 7.0, 6.8, 5.5, 8.4]
tickets = [500000, 650000, 480000, 400000, 800000]

fig, ax = plt.subplots(figsize=(10, 6))
ax.scatter(ratings, tickets, color='orange', s=100)

for i in range(len(movies)):
    ax.annotate(movies[i], (ratings[i], tickets[i]), textcoords="offset points", xytext=(0, 10), ha='center')

ax.set_xlabel('IMDB Rating')
ax.set_ylabel('Tickets Sold')
ax.set_title('Bollywood Movies: Ratings vs Tickets Sold')
```

# question 5

```python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
sales = [120, 110, 130, 140, 150, 160, 145, 155, 175, 450, 190, 210]

fig, ax = plt.subplots(figsize=(12, 6))
ax.plot(months, sales, marker='o', color='purple', linewidth=2)

max_sales = max(sales)
max_index = sales.index(max_sales)

ax.annotate('Big Billion Days', 
            xy=(months[max_index], max_sales), 
            xytext=(max_index - 1.5, max_sales + 20),
            arrowprops=dict(facecolor='black', shrink=0.05, width=1.5, headwidth=8),
            fontsize=12, fontweight='bold')

ax.set_xlabel('Month')
ax.set_ylabel('Monthly Sales')
ax.set_title('Flipkart Monthly Sales')
```
