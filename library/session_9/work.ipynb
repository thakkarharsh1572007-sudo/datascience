# question 2

```python
df = pd.read_csv("flipkart_order_history_10k.csv")
df['order_date'] = pd.to_datetime(df['order_date'])
# year, month, and weekday
df['year'] = df['order_date'].dt.year
df['date'] = df['order_date'].dt.date
df['day'] = df['order_date'].dt.day_name()
df.head(10)
```

```python
df = pd.read_csv('instagram_posts_dataset.csv')
print(df.head())
df['posted_at_utc'] = pd.to_datetime(df['posted_at_utc']).dt.tz_localize('UTC')
df['posted_at_ist'] = df['posted_at_utc'].dt.tz_convert('Asia/Kolkata')
df.head()
```
