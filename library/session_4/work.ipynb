# question 5

```python
import pandas as pd

# 1. Simulate a small sample semicolon-separated CSV file
csv_data = """Transaction_ID;Date;Amount;Status
TXN001;2026-06-01;500.0;Success
TXN002;2026-06-02;;Failed
TXN003;;150.0;Success
TXN004;2026-06-04;1200.0;
"""

file_name = "paytm_transactions.csv"
with open(file_name, "w") as f:
    f.write(csv_data)

# 2. Read the semicolon-separated CSV file
# Using sep=';' ensures pandas parses the columns correctly instead of reading everything as one column
df = pd.read_csv(file_name, sep=";")

# 3. Detect missing values per column
missing_counts = df.isnull().sum()

# 4. Filter and print columns that have nulls (> 0)
print("--- Missing Values Detection Report ---")
columns_with_nulls = missing_counts[missing_counts > 0]

if not columns_with_nulls.empty:
    for col, count in columns_with_nulls.items():
        print(f"Column '{col}' has {count} missing value(s).")
else:
    print("No missing values found in the file.")
```
